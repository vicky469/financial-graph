# SEC Registrant Analysis

Analyze unique registrant names from SEC metadata

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Configure pandas display
pd.set_option('display.max_rows', 100)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
```

## Load Data

```python
# Path to CSV file
csv_path = Path('../output/registrant_metadata_2025.csv')

# Load the CSV
df = pd.read_csv(csv_path)

print(f"Total rows: {len(df):,}")
print(f"\nColumns: {list(df.columns)}")
df.head()
```

## Unique Registrant Names

```python
# Count unique registrant names
unique_registrants = df['registrant_name'].nunique()

print(f"Total filings: {len(df):,}")
print(f"Unique registrant names: {unique_registrants:,}")
print(f"Average filings per registrant: {len(df) / unique_registrants:.2f}")
```

## Top Registrants by Filing Count

```python
# Top 20 most active registrants
top_registrants = df['registrant_name'].value_counts().head(20)

print("Top 20 Most Active Registrants (by number of filings):")
print("=" * 80)
for i, (name, count) in enumerate(top_registrants.items(), 1):
    print(f"{i:2d}. {name[:60]:<60} {count:>6,} filings")

top_registrants
```

## Filing Type Distribution

```python
# Count by form type
form_counts = df['form_type'].value_counts()

print("Form Type Counts (all):")
print("=" * 60)
for form, count in form_counts.items():
    pct = (count / len(df)) * 100
    print(f"{form:<15} {count:>8,} ({pct:>5.2f}%)")

form_counts
```

## CIK Analysis

```python
# Unique CIKs
unique_ciks = df['cik'].nunique()

print(f"Unique CIKs: {unique_ciks:,}")
print(f"Unique Registrant Names: {unique_registrants:,}")
print(f"\nDifference: {abs(unique_ciks - unique_registrants):,}")
print("(Same company may have multiple names due to mergers, name changes, etc.)")
```

## Companies with Multiple Names (Name Changes)

Since there are 151,698 unique CIKs but 152,402 unique registrant names, some companies have filed under multiple names (due to mergers, acquisitions, rebranding, etc.). Let's identify these companies.

```python
# Group by CIK and find CIKs with multiple names
cik_names = df.groupby('cik')['registrant_name'].apply(lambda x: x.unique()).reset_index()
cik_names['name_count'] = cik_names['registrant_name'].apply(len)

# Filter to CIKs with multiple names
multiple_names = cik_names[cik_names['name_count'] > 1].sort_values('name_count', ascending=False)

print(f"CIKs with multiple names: {len(multiple_names):,}")
print(f"CIKs with single name: {len(cik_names[cik_names['name_count'] == 1]):,}")
print(f"\nTop 20 companies with most name variations:")
print("=" * 80)

for idx, row in multiple_names.head(20).iterrows():
    print(f"\nCIK {row['cik']} ({row['name_count']} names):")
    for name in row['registrant_name']:
        print(f"  • {name}")
```

```python
# Analyze name changes over time and identify primary (current) name
# Primary name = name used in most recent filing

def show_name_timeline(cik_value):
    """Show how a company's name changed over time"""
    company_filings = df[df['cik'] == cik_value].sort_values('filing_date')
    
    # Get unique names with first and last filing dates
    name_timeline = company_filings.groupby('registrant_name').agg({
        'filing_date': ['min', 'max', 'count']
    }).reset_index()
    name_timeline.columns = ['registrant_name', 'first_filing', 'last_filing', 'filing_count']
    name_timeline = name_timeline.sort_values('first_filing')
    
    # Identify primary name (most recent filing)
    most_recent_filing = company_filings.iloc[-1]
    primary_name = most_recent_filing['registrant_name']
    
    print(f"\nCIK {cik_value} - Name Timeline:")
    print(f"PRIMARY (Current): {primary_name}")
    print("=" * 100)
    for _, row in name_timeline.iterrows():
        is_primary = "[PRIMARY]" if row['registrant_name'] == primary_name else "[ALIAS]  "
        print(f"{is_primary} {row['registrant_name']:<55} {row['first_filing']} to {row['last_filing']} ({row['filing_count']} filings)")
    
    return name_timeline, primary_name

# Show examples for top companies with multiple names
for cik in multiple_names.head(5)['cik'].values:
    show_name_timeline(cik)
```

```python
# Distribution of name count
name_count_dist = cik_names['name_count'].value_counts().sort_index()

print("Distribution of Name Counts per CIK:")
print("=" * 60)
for count, freq in name_count_dist.head(10).items():
    pct = (freq / len(cik_names)) * 100
    print(f"{count} name(s): {freq:>8,} CIKs ({pct:>5.2f}%)")

if len(name_count_dist) > 10:
    print(f"...")
    print(f"Max names for single CIK: {name_count_dist.index.max()}")
```

## Primary Name Determination Logic

```python
# Explain the logic for determining primary name
print("How to Determine Primary (Current) Name for a CIK:")
print("=" * 80)
print("\n1. PRIMARY NAME = Name used in the MOST RECENT filing")
print("   - This represents the company's current legal name")
print("   - Most reliable indicator of current identity")
print("\n2. ALIAS NAMES = All other historical names")
print("   - Previous legal names (before rebranding/merger)")
print("   - Should be stored in company.aliases[] array")
print("\n3. Example Timeline:")
print("   2020-2021: 'Facebook, Inc.' [ALIAS]")
print("   2021-2025: 'Meta Platforms, Inc.' [PRIMARY]")
print("   → Primary: 'Meta Platforms, Inc.'")
print("   → Aliases: ['Facebook, Inc.']")
print("\n4. Edge Cases:")
print("   - If names overlap in time: Use most recent filing date")
print("   - If multiple names on same date: Use name with most filings on that date")
print("   - If still tied: Alphabetically first (deterministic)")

# Show statistics on primary vs alias names
print("\n" + "=" * 80)
print("Statistics:")
print(f"  Total CIKs with multiple names: {len(multiple_names):,}")
print(f"  Total name records: {multiple_names['name_count'].sum():,}")
print(f"  Primary names (1 per CIK): {len(multiple_names):,}")
print(f"  Alias names: {multiple_names['name_count'].sum() - len(multiple_names):,}")
print(f"  Average aliases per CIK: {(multiple_names['name_count'].sum() - len(multiple_names)) / len(multiple_names):.2f}")
```

```python
# Export CIKs with multiple names to CSV for further analysis
output_path = Path('../output/registrant_metadata_2025_multiple_names.csv')

# Create detailed export with all names per CIK, marking primary name
export_data = []
for _, row in multiple_names.iterrows():
    cik = row['cik']
    names = row['registrant_name']
    
    # Get filing dates for each name
    company_filings = df[df['cik'] == cik]
    
    # Identify primary name (most recent filing)
    most_recent_filing = company_filings.sort_values('filing_date').iloc[-1]
    primary_name = most_recent_filing['registrant_name']
    
    for name in names:
        name_filings = company_filings[company_filings['registrant_name'] == name]
        export_data.append({
            'cik': cik,
            'registrant_name': name,
            'is_primary': name == primary_name,
            'first_filing_date': name_filings['filing_date'].min(),
            'last_filing_date': name_filings['filing_date'].max(),
            'filing_count': len(name_filings),
            'total_names_for_cik': row['name_count']
        })

export_df = pd.DataFrame(export_data)
export_df = export_df.sort_values(['cik', 'is_primary', 'first_filing_date'], ascending=[True, False, True])
export_df.to_csv(output_path, index=False)

print(f"\nExported {len(export_df):,} name records for {len(multiple_names):,} CIKs to:")
print(f"{output_path}")
print(f"\nPrimary names: {export_df['is_primary'].sum():,}")
print(f"Alias names: {(~export_df['is_primary']).sum():,}")
print(f"\nSample of export (showing primary names first):")
export_df.head(20)
```

## Notable Examples: Well-Known Name Changes

```python
# Search for well-known companies that changed names
notable_searches = [
    ('Meta', 'Facebook'),
    ('Alphabet', 'Google'),
    ('X Corp', 'Twitter'),
    ('Warner Bros', 'Time Warner'),
]

print("Notable Name Changes:")
print("=" * 100)

for search1, search2 in notable_searches:
    # Find CIKs that have both names
    matches1 = df[df['registrant_name'].str.contains(search1, case=False, na=False)]['cik'].unique()
    matches2 = df[df['registrant_name'].str.contains(search2, case=False, na=False)]['cik'].unique()
    
    common_ciks = set(matches1) & set(matches2)
    
    if common_ciks:
        print(f"\nFound CIK(s) with both '{search1}' and '{search2}':")
        for cik in common_ciks:
            show_name_timeline(cik)
    else:
        print(f"\nNo CIK found with both '{search1}' and '{search2}'")
        if matches1.any():
            print(f"  Found {len(matches1)} CIK(s) with '{search1}'")
        if matches2.any():
            print(f"  Found {len(matches2)} CIK(s) with '{search2}'")
```

## Same Name, Multiple CIKs

The reverse case: same registrant name used by multiple CIKs. This could indicate:
- Different legal entities with similar/identical names
- Subsidiaries with same name in different jurisdictions
- Data quality issues or name collisions

```python
# Group by registrant name and find names with multiple CIKs
name_ciks = df.groupby('registrant_name')['cik'].apply(lambda x: x.unique()).reset_index()
name_ciks['cik_count'] = name_ciks['cik'].apply(len)

# Filter to names with multiple CIKs
multiple_ciks = name_ciks[name_ciks['cik_count'] > 1].sort_values('cik_count', ascending=False)

print(f"Names with multiple CIKs: {len(multiple_ciks):,}")
print(f"Names with single CIK: {len(name_ciks[name_ciks['cik_count'] == 1]):,}")
print(f"\nTop 20 names with most CIK variations:")
print("=" * 80)

for idx, row in multiple_ciks.head(20).iterrows():
    print(f"\n'{row['registrant_name']}' ({row['cik_count']} CIKs):")
    for cik in row['cik']:
        filing_count = len(df[(df['registrant_name'] == row['registrant_name']) & (df['cik'] == cik)])
        print(f"  • CIK {cik} - {filing_count} filings")
```

```python
# Distribution of CIK count per name
cik_count_dist = name_ciks['cik_count'].value_counts().sort_index()

print("Distribution of CIK Counts per Name:")
print("=" * 60)
for count, freq in cik_count_dist.head(10).items():
    pct = (freq / len(name_ciks)) * 100
    print(f"{count} CIK(s): {freq:>8,} names ({pct:>5.2f}%)")

if len(cik_count_dist) > 10:
    print(f"...")
    print(f"Max CIKs for single name: {cik_count_dist.index.max()}")
```

```python
# Analyze timeline for names with multiple CIKs
def show_cik_timeline(name):
    """Show how different CIKs used the same name over time"""
    name_filings = df[df['registrant_name'] == name].sort_values('filing_date')
    
    # Get unique CIKs with first and last filing dates
    cik_timeline = name_filings.groupby('cik').agg({
        'filing_date': ['min', 'max', 'count']
    }).reset_index()
    cik_timeline.columns = ['cik', 'first_filing', 'last_filing', 'filing_count']
    cik_timeline = cik_timeline.sort_values('first_filing')
    
    print(f"\n'{name}' - CIK Timeline:")
    print("=" * 100)
    for _, row in cik_timeline.iterrows():
        print(f"CIK {row['cik']:<15} {row['first_filing']} to {row['last_filing']} ({row['filing_count']} filings)")
    
    return cik_timeline

# Show examples for top names with multiple CIKs
for name in multiple_ciks.head(5)['registrant_name'].values:
    show_cik_timeline(name)
```

```python
# Export names with multiple CIKs to CSV
output_path = Path('../output/registrant_metadata_2025_multiple_ciks.csv')

# Create detailed export with all CIKs per name
export_data = []
for _, row in multiple_ciks.iterrows():
    name = row['registrant_name']
    ciks = row['cik']
    
    # Get filing dates for each CIK
    name_filings = df[df['registrant_name'] == name]
    
    for cik in ciks:
        cik_filings = name_filings[name_filings['cik'] == cik]
        export_data.append({
            'registrant_name': name,
            'cik': cik,
            'first_filing_date': cik_filings['filing_date'].min(),
            'last_filing_date': cik_filings['filing_date'].max(),
            'filing_count': len(cik_filings),
            'total_ciks_for_name': row['cik_count']
        })

export_df = pd.DataFrame(export_data)
export_df = export_df.sort_values(['registrant_name', 'first_filing_date'])
export_df.to_csv(output_path, index=False)

print(f"\nExported {len(export_df):,} CIK records for {len(multiple_ciks):,} names to:")
print(f"{output_path}")
print(f"\nSample of export:")
export_df.head(20)
```
