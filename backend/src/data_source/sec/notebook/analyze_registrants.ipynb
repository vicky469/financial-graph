# SEC Registrant Analysis

Analyze unique registrant names from SEC metadata

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Configure pandas display
pd.set_option('display.max_rows', 100)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
```

## Load Data

```python
# Path to CSV file
csv_path = Path('../output/registrant_metadata_2025.csv')

# Load the CSV
df = pd.read_csv(csv_path)

print(f"Total rows: {len(df):,}")
print(f"\nColumns: {list(df.columns)}")
df.head()
```

## Unique Registrant Names

```python
# Count unique registrant names
unique_registrants = df['registrant_name'].nunique()

print(f"Total filings: {len(df):,}")
print(f"Unique registrant names: {unique_registrants:,}")
print(f"Average filings per registrant: {len(df) / unique_registrants:.2f}")
```

## Top Registrants by Filing Count

```python
# Top 20 most active registrants
top_registrants = df['registrant_name'].value_counts().head(20)

print("Top 20 Most Active Registrants (by number of filings):")
print("=" * 80)
for i, (name, count) in enumerate(top_registrants.items(), 1):
    print(f"{i:2d}. {name[:60]:<60} {count:>6,} filings")

top_registrants
```

## Filing Type Distribution

```python
# Count by form type
form_counts = df['form_type'].value_counts()

print("Form Type Counts (all):")
print("=" * 60)
for form, count in form_counts.items():
    pct = (count / len(df)) * 100
    print(f"{form:<15} {count:>8,} ({pct:>5.2f}%)")

form_counts
```

## CIK Analysis

```python
# Unique CIKs
unique_ciks = df['cik'].nunique()

print(f"Unique CIKs: {unique_ciks:,}")
print(f"Unique Registrant Names: {unique_registrants:,}")
print(f"\nDifference: {abs(unique_ciks - unique_registrants):,}")
print("(Same company may have multiple names due to mergers, name changes, etc.)")
```

## Quarterly Breakdown

```python
# Filings by quarter
quarterly = df['source_quarter'].value_counts().sort_index()

print("Filings by Quarter:")
print("=" * 40)
for quarter, count in quarterly.items():
    print(f"{quarter}: {count:>10,} filings")

quarterly
```

## Sample Companies (Search by Name)

```python
# Search for specific companies
search_term = "APPLE"  # Change this to search for different companies

matches = df[df['registrant_name'].str.contains(search_term, case=False, na=False)]

print(f"Companies matching '{search_term}':")
print("=" * 80)
print(f"Found {len(matches):,} filings from {matches['registrant_name'].nunique()} unique entities")
print("\nUnique registrant names:")
for name in matches['registrant_name'].unique()[:10]:
    count = len(matches[matches['registrant_name'] == name])
    cik = matches[matches['registrant_name'] == name]['cik'].iloc[0]
    print(f"  • {name} (CIK: {cik}) - {count} filings")

matches.head(10)
```

## Data Quality Checks

```python
# Check for missing values
print("Missing Values:")
print("=" * 40)
print(df.isnull().sum())

# Check for duplicates
duplicates = df.duplicated(subset=['cik', 'accession_number_nodashes']).sum()
print(f"\nDuplicate filings (CIK + Accession): {duplicates:,}")
```

## Export Summary Statistics

```python
# Create summary dataframe
summary = {
    'Metric': [
        'Total Filings',
        'Unique Registrant Names',
        'Unique CIKs',
        'Unique Form Types',
        'Date Range'
    ],
    'Value': [
        f"{len(df):,}",
        f"{unique_registrants:,}",
        f"{unique_ciks:,}",
        f"{df['form_type'].nunique():,}",
        f"{df['filing_date'].min()} to {df['filing_date'].max()}"
    ]
}

summary_df = pd.DataFrame(summary)
print("\nSummary Statistics:")
print("=" * 60)
print(summary_df.to_string(index=False))

summary_df
```
